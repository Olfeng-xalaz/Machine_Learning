## Abschnitt 5 – Erste Visualisierung

Wir erstellen ein einfaches Liniendiagramm.  
Später werden wir im Kurs häufiger Streudiagramme verwenden.


```python
plt.figure(figsize=(6, 4))
plt.plot(df["Lernzeit_Stunden"], df["Punktzahl"], marker="o")
plt.xlabel("Lernzeit in Stunden")
plt.ylabel("Punktzahl")
plt.title("Lernzeit und Punktzahl")
plt.grid(True)
plt.show()
```

```python
# Diagramm mit den geänderten Werten neu zeichnen
plt.figure(figsize=(6, 4))
plt.plot(df["Lernzeit_Stunden"], df["Punktzahl"], marker="o")
plt.xlabel("Lernzeit in Stunden (geändert)")
plt.ylabel("Punktzahl")
plt.title("Lernzeit und Punktzahl (mit geänderten Lernzeiten)")
plt.grid(True)
plt.show()
```

1. Was ist der Unterschied zwischen einer Python-Liste und einem NumPy-Array?

Python-Liste: Eine Python-Liste ist eine vielseitige, geordnete und veränderliche Sammlung von Elementen, die verschiedene Datentypen enthalten können (z.B. Zahlen, Strings, andere Listen). Sie ist sehr flexibel, aber nicht optimiert für numerische Operationen.
NumPy-Array: Ein NumPy-Array ist eine Datenstruktur, die speziell für effiziente numerische Berechnungen entwickelt wurde. Alle Elemente in einem NumPy-Array müssen vom gleichen Datentyp sein. Dies ermöglicht es NumPy, Operationen (z.B. mathematische Berechnungen) viel schneller auszuführen als mit Python-Listen, besonders bei großen Datenmengen. NumPy-Arrays sind auch speichereffizienter.

2. Warum ist ein DataFrame für Datensätze oft praktischer als eine einfache Liste?

  Ein DataFrame (von der pandas-Bibliothek) ist eine tabellarische Datenstruktur mit Zeilen und Spalten, die man sich wie eine Tabelle in einer Datenbank oder ein Tabellenblatt in Excel vorstellen kann. Es ist praktischer als eine einfache Liste, weil:
  
  Strukturierte Daten: Es ermöglicht die Speicherung heterogener Daten (verschiedene Datentypen in verschiedenen Spalten) in einer organisierten, spaltenbasierten Struktur. Eine Liste wäre eine flache Sammlung, die diese Spaltenlogik nicht direkt abbildet.

Beschriftung und Indexierung: Jede Spalte hat einen Namen und jede Zeile einen Index, was den Zugriff, die Auswahl und die Bearbeitung von Daten intuitiver und einfacher macht als bei einer einfachen Liste, bei der man sich auf numerische Indizes verlassen müsste.


Leistungsstarke Operationen: DataFrames bieten eingebaute Funktionen und Methoden für Datenbereinigung, -analyse, -transformation und -aggregation (z.B. Filtern, Gruppieren, Zusammenführen), die speziell für tabellarische Daten optimiert sind und bei Listen manuell implementiert werden müssten.

Kompatibilität: Sie sind das Standardformat für viele Machine-Learning-Bibliotheken und lassen sich leicht aus verschiedenen Datenquellen (CSV, Excel, Datenbanken) laden und speichern.
